```python
import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
ASSETS = ['AAPL', 'MSFT', 'NVDA', 'JPM', 'SPY', 'GLD']

df = yf.download(ASSETS, start="2018-01-01", end="2024-01-01")['Close']
df.dropna(inplace=True)
df.to_csv("../data/prices.csv")
print(df.shape)
df.head()
```
